## Alterações metodológicas

A produção em `t` foi tratada como energia do intervalo `[t,t+1h)`. A irradiância foi alinhada usando `GHI(t+1h)` e o zênite recalculado no ponto médio (`t+30 min`). Temperatura e vento permaneceram em `t`. Os 185 zeros virtuais acordados foram excluídos sem inspeção individual. A última linha, sem GHI alinhada disponível, foi removida da população elegível.

## Limites de interpretação

A comparação controlada é a pareada no conjunto comum. A metodologia V3 mudou simultaneamente o alinhamento da GHI, o instante do zênite e a regra dos zeros virtuais; o efeito não pode ser atribuído a uma única alteração. LOMO é retrospectivo, com meteorologia histórica, e não comprova desempenho day-ahead usando previsões disponíveis às 21h. O teste cronológico fevereiro–março foi previamente inspecionado e é exploratório.
